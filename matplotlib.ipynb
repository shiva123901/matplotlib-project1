```python
x=np.linspace(0,3*np.pi,100)
# Defining the wave
sin_y=np.sin(x)
cos_y=np.cos(x)
#Showing the style
plt.style.use("seaborn-v0_8")
#ploting the wave
plt.plot(x,sin_y,label="Sine")
plt.plot(x,cos_y,label="Cosine")
plt.legend(loc="upper right")
plt.show()
```

```python
x=np.linspace(0,3*np.pi,100)
# Defining the wave
sin_y=np.sin(x)
cos_y=np.cos(x)
# Definiong the size
plt.figure(figsize=(18,5))
#Showing the style
plt.style.use("grayscale")
#ploting the wave
plt.plot(x,sin_y,label="Sine")
plt.plot(x,cos_y,label="Cosine")
plt.legend(loc="upper right")
plt.show()
```

#### 3. saving the figure

```python
x=np.linspace(0,4*np.pi,100)
# Defining the wave
sin_y=np.sin(x)
cos_y=np.cos(x)
# Definiong the size
plt.figure(figsize=(18,5))
#Showing the style
plt.style.use("grayscale")
#ploting the wave
plt.plot(x,sin_y,label="Sine")
plt.plot(x,cos_y,label="Cosine")
plt.legend(loc="upper right")
plt.title("Sine and Cosine wave")
fig=plt.figure()

plt.savefig('Sine & Cosine.png')
```

#### Multiple line plot

```python
x=np.linspace(0,4*np.pi,100)
# Defining the wave
sin_y=np.sin(x)
cos_y=np.cos(x)
# Definiong the size
plt.figure(figsize=(18,5))
#Showing the style
plt.style.use("ggplot")
#ploting the wave
plt.plot(x,sin_y,"*",color='blue',linewidth=1,label="Sine",)
plt.plot(x,cos_y,"-",color="green",linewidth=1,label="Cosine")
plt.legend(loc="upper right")
plt.title("Sine and Cosine wave",color="red")
plt.show()
```

## SCATTER PLOT

```python
n=500
x=np.random.randn(n)
y=np.random.randn(n)

plt.scatter(x,y,c="red",alpha=0.7,marker=">")

plt.title("Scatterplot",color="red")
plt.show()
```

```python
n=500
x1=np.random.randn(n)
y1=np.random.randn(n)

x2=np.random.randn(n)+2
y2=np.random.randn(n)


plt.scatter(x1,y1, c="red", alpha=0.7, s=100, label="Group1",marker=">")
plt.scatter(x2,y2, c="blue", alpha=0.7, s=100, label="Group2")
plt.legend()

plt.title("Scatterplot",color="red")
plt.show()
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
x = iris.data
y = iris.target

plt.scatter(x[:,2],x[:,3], c = 'blue', s =50, alpha = .7)

plt.xlabel(iris['feature_names'][2])
plt.ylabel(iris['feature_names'][3])

plt.show()
```

```python
rng=np.random.RandomState(0)

x= rng.randn(100)
y= rng.randn(100)
color=rng.rand(100)
size=rng.rand(100)*1000
plt.scatter(x,y,c=color,s=size,alpha=0.95,cmap='viridis',edgecolors='black')
plt.title("Random Scatter plot",color='red')
plt.xlabel("X value")
plt.ylabel("Y value")
plt.colorbar()
plt.show()
```

### Histogram and binning

```python
data=np.random.randn(1000)
plt.style.use("Solarize_Light2")
plt.hist(data)

plt.show()
```

```python
data=np.random.randn(10000000)
plt.style.use("Solarize_Light2")
plt.hist(data, bins=10000,edgecolor='red',color='lightblue',alpha=0.8)
plt.xlabel("Value",color="red")
plt.ylabel("Frequency",color="red")
plt.title("Random Histogram",color="red")

plt.show()
```

```python
data=np.random.randn(100000)
plt.style.use("Solarize_Light2")
plt.hist(data, bins=10000,edgecolor='red',color='lightblue',alpha=0.8,histtype="stepfilled")
plt.xlabel("Value",color="red")
plt.ylabel("Frequency",color="red")
plt.title("Random Histogram",color="red")

plt.show()
```

```python
x1=np.random.randn(100)
x2=np.random.randn(100)+2
x3=np.random.randn(100)+4

plt.style.use("Solarize_Light2")
plt.hist(x1, bins=20,alpha=0.8)
plt.hist(x2, bins=20,alpha=0.8)
plt.hist(x3, bins=20,alpha=0.8)
plt.xlabel("Value",color="red")
plt.ylabel("Frequency",color="red")
plt.title("Random Histogram",color="red")

plt.show()
```

### 2D Histogram

```python
x=np.random.randn(1000)
y=np.random.randn(1000)

plt.style.use("Solarize_Light2")
plt.hist2d(x,y, bins=20,cmap="Blues")
plt.xlabel("Value",color="red")
plt.ylabel("Frequency",color="red")
plt.title("Random 2D Histogram",color="red")

plt.show()
```
